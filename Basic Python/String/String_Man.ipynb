```python
#Upper
txt = "Hello my friend"
print(txt.upper())
```

```python
#Lower
txt = "HELLO MY FRIEND"
print(txt.lower())
```

```python
#Capitalise 
txt = "hello my friend"
print(txt.capitalize())
```

```python
#Split
txt = ("welcome,to,the,jungle")
print(txt.split(","))
```

```python
#startswith
txt = "Hello my friend"
print(txt.startswith("Hello"))
```

```python
#endswith
txt = "hello my friend"
print(txt.endswith("friend"))
```

```python
#index
txt = "hello my friend"
print(txt.index("my"))
```

```python
#isnumeric
txt = "5050"
print(txt.isnumeric())
```

```python
#isalpha
txt = "Soham"
print(txt.isalpha())
```

```python
#alphanum
txt = "Soham13"
print(txt.isalnum())
```

```python
# extract substring
txt = "rajesh40"
new_value = ""

for i in txt:
    if i.isalpha():
        new_value = new_value + i

print(new_value)
```
