```python
#Add
S1 = {"apple", "banana", "mango"}
S1.add("orange")
S1
```

```python
#Update
S1 = {"apple", "banana", "mango"}
S1.update(["watermelon", "pineapple"])
S1
```

```python
#remove
S1 = {"apple", "banana", "mango"}
S1.remove("banana")
S1
```

```python
#discard()
S1 = {"apple", "banana", "mango"}
S1.discard("mango")
S1
```

```python
#clear()
S1 = {"apple", "banana", "mango"}
S1.clear()
S1
```

```python
#union
s1 = {1, 2, 3}
s2 = {4, 5, 6}
s3 = s1.union(s2)
s3
```

```python
#intersection
s1 = {2, 4, 6, 8, 10, 12, 14}
s2 = {12, 14, 16, 18, 20}
s3 = s1.intersection(s2)
s3

```

```python
#difference
s1 = {5, 10, 15, 20, 25}
s2 = {20, 25, 30, 35, 40}
s3 = s1.difference(s2)
s3
```

```python
#symmetric_difference
s1 = {10, 20, 30, 40, 50, 60}
s2 = {50, 60, 70, 80, 90, 100}
s3 = s1.symmetric_difference(s2)
s3
```
