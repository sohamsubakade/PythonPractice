```python
#Create Dict 
Dict_01 = {"name" : "Rohan", "age" : 23}
Dict_01
```

```python
#Using Constructor
Dict = dict(name="Rohan", age=23)
Dict
```

```python
#using keys 
dict = {"name" : "Rohan", "age" : 24, "city" : "Pune"}
for i in dict:
    print(i)
```

```python
#using keys 1.0
Dict = {"name" : "Rohan", "age" : 24, "city" : "Pune"}
for i in dict.keys():
    print(i)
```

```python
Dict = {"name" : "Rohan", "age" : 24, "city" : "Pune"}
for i in dict.keys():
   print(dict[i])
```

```python
#using values
Dict = {"name" : "Rohan", "age" : 24, "city" : "Pune"}
for i in dict.values():
    print(i)
```

```python
#using items
Dict = {"name" : "Rohan", "age" : 24, "city" : "Pune"}
for i, j in dict.items():
    print(i,j)
    
```

```python
#Remove using pop
Dict = {"name" : "Rohan", "age" : 24, "city" : "Pune"}
dict.pop("city")
dict
```

```python
#Remove using popitem
Dict = {"name" : "Rohan", "age" : 24, "city" : "Pune"}
Dict.popitem()
Dict
```

```python
#Remove usind del
Dict = {"name" : "Rohan", "age" : 24, "city" : "Pune"}
del Dict["name"]
Dict
```

```python
#Update 
Dict = {"name" : "Rohan", "age" : 24, "city" : "Pune"}
Dict.update({"Pincode" : 411038, "Area" : "Kothrud"})
Dict
```

```python
#Copy
Dict = {"name" : "Rohan", "age" : 24, "city" : "Pune"}
dict_2 = dict.copy()
dict_2
```

```python
#Merge
dict = {"name" : "Rohan", "age" : 24, "city" : "Pune"}
dict_02 = {"Pincode" : 411038, "Area" : "Kothrud"}

dict_03 = {**dict,**dict_02}
print(dict_03)
```

```python
#Join
dict = {"name" : "Rohan", "age" : 24, "city" : "Pune"}
dict_02 = {"Pincode" : 411038, "Area" : "Kothrud"}
dict.update(dict_02)
dict
```
